## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

1. Search Volume vs. Impressions: Pearson correlation between search_volume and impressions_90d is 0.001, proving raw volume alone does not drive traffic.

2. CTR Collapse: Mean Click-Through-Rate drops sharply from 35.5% on Page 1 to 14.2% on Pages 3–5, highlighting the urgency of flagging striking-distance decay.

3. Word Count Myth: Pages trending down have a median word count of 2,909, nearly identical to pages trending up (2,848), showing length alone is not a proxy for freshness.

```python
# This cell is for CODE (numbers, a query, a check).
import pandas as pd

# Load dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 1. Search Volume vs Impressions Correlation
corr = df['search_volume'].corr(df['impressions_90d'])
print(f"1. Search Volume vs Impressions Correlation: {corr:.3f}")

# 2. Average CTR by Position Group (multiply by 100 for percentage)
page_1_ctr = df[df['avg_position'] <= 10]['ctr'].mean() * 100
page_3_5_ctr = df[(df['avg_position'] > 20) & (df['avg_position'] <= 50)]['ctr'].mean() * 100
print(f"2. Page 1 CTR: {page_1_ctr:.1f}% | Page 3-5 CTR: {page_3_5_ctr:.1f}%")

# 3. Median Word Count by Trend Direction
declining_wc = df[df['trend_direction'] == 'down']['word_count'].median()
growing_wc = df[df['trend_direction'] == 'up']['word_count'].median()
print(f"3. Median Word Count — Declining: {declining_wc:.0f} | Growing: {growing_wc:.0f}")
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```
